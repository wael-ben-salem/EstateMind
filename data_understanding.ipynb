```python
data = pd.read_json('datasets/monbien_tn/data.json')
df_monbien = json_normalize(data.to_dict('records'))

df_monbien.info()

df_monbien.describe(include='all').T
summary = pd.DataFrame({
    'dtype': df_monbien.dtypes,
    'non_null_count': df_monbien.notnull().sum(),
    'null_count': df_monbien.isnull().sum(),
    'completeness_%': df_monbien.notnull().sum() / len(df_monbien) * 100,
})

duplicates = []
fully_duped_cont = []

for col in df_monbien.columns:
    try:
        # Get the most common value and its frequency
        counts = df_monbien[col].value_counts(dropna=False)
        most_common_freq = counts.iloc[0]
        duplicates.append(most_common_freq / len(df_monbien) * 100)  # % of most common value

        if most_common_freq / len(df_monbien) > 0.99:
            fully_duped_cont.append(counts.index[0])
        else:
            fully_duped_cont.append(None)

    except Exception as e:
        duplicates.append(None)
        fully_duped_cont.append(None)
        print(f"Cannot compute duplicates for column {col}: {e}")

summary['duplicates_in_column'] = duplicates
summary['fully_duplicated_content'] = fully_duped_cont

summary
df_monbien
```

```python
import ast

def add_features_to_caracteristiques(row):
    # Ensure caracteristiques is a list
    if isinstance(row['caracteristiques'], str):
        try:
            features_list = ast.literal_eval(row['caracteristiques'])
        except:
            features_list = []
    elif isinstance(row['caracteristiques'], list):
        features_list = row['caracteristiques']
    else:
        features_list = []

    # Add features if value == 1
    if row.get('clim') == 1:
        features_list.append("clim")

    if row.get('ascenseur') == 1:
        features_list.append("ascenseur")

    if row.get('chauffage') == 1:
        features_list.append("chauffage")

    if row.get('jardin') == 1:
        features_list.append("jardin")

    return list(set(features_list))  # remove duplicates if any


df_tecnocasa_std['caracteristiques'] = df_tecnocasa_std.apply(add_features_to_caracteristiques, axis=1)

df_tecnocasa_std['caracteristiques'].value_counts()
```

Quick Pre-clean steps

Numeric conversion:

prix, surface, chambres, annee_constr, etage → convert to numeric (some are objects).

```python
import pandas as pd
import numpy as np

df = df_tecnocasa_std.copy()  # keep original safe

# ----------------------------
# 1. Numeric conversion
# ----------------------------
# Convert to numeric
# basic numeric columns
for col in ['annee_constr', 'prix', 'surface']:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# features that may have non-numeric chars
# Example: "3 pieces" -> 3, "4 pièces" -> 4
if 'pieces' in df.columns:
    df['pieces'] = df['pieces'].astype(str).str.extract('(\d+)').astype('Int64')

if 'etage' in df.columns:
    manual_mapping = {
    'RDC': 0,
    'Terre': 0,
    'Sous-sol': -1,
    'Sot': -1,
    'Bas': 1,
    'Moyen': 2,
    'Elevé': 3,
    'Dernier': np.nan,  # optional: choose max floor or leave NaN
    'Grenier': np.nan   # optional: choose a value
}

def process_floor(val):
    val_str = str(val).strip()
    
    # If empty, return NaN
    if val_str == '':
        return np.nan
    
    # Try extracting first number
    num = pd.Series([int(x) for x in val_str if x.isdigit()])
    if not num.empty:
        return int(num.iloc[0])
    
    # If no number, check mapping
    return manual_mapping.get(val_str, np.nan)  # default NaN if unknown

# Apply function
df['etage'] = df['etage'].apply(process_floor).astype('Int64')
print(df['etage'].unique())



# ----------------------------
# 2. Binary features: convert to 0/1
# ----------------------------
binary_cols = ['clim', 'ascenseur', 'chauffage', 'jardin']
for col in binary_cols:
    if col in df.columns:
        df[col] = df[col].apply(lambda x: 1 if str(x).strip().lower() in ['yes','oui','true','1'] else 0)
        df[col] = df[col].fillna(0).astype(int)


# ----------------------------
# 3. Fill NaNs
# ----------------------------
# Numeric columns
#numeric_cols = ['prix', 'surface', 'pieces', 'etage', 'annee_constr']
#for col in numeric_cols:
#    if col in df.columns:
#        df[col] = df[col].fillna(0)

# Categorical columns
categorical_cols = ['ville','contrat','standing','region','type','transport']
for col in categorical_cols:
    if col in df.columns:
        df[col] = df[col].fillna('unknown').astype(str).str.strip()

# ----------------------------
# 4. Feature columns already numeric, ensure int
# ----------------------------
feature_cols = ['aire_jeux','alarme','ascenseur_bat','balcon','brique','cafeteria','divise',
                'doublure','enduction','entree_indep','escalier_ext','four','garage','gardien',
                'interphone','jardin_ext','parking_pub','pergola_abris','place_parking','plafond',
                'potager','poutres','recouvrement','rustique','reception','anti_incendie','terrasse',
                'videophone','cour_interieure']
for col in feature_cols:
    if col in df.columns:
        df[col] = df[col].fillna(0).astype(int)

# ----------------------------
# 5. Ensure datetime
# ----------------------------
if 'last_published_at' in df.columns:
    df['last_published_at'] = pd.to_datetime(df['last_published_at'], errors='coerce')

# ----------------------------
# Done
# ----------------------------

df_tecnocasa_std = df.copy()  # save cleaned version

print("Pre-clean done! Sample:")
df
```
