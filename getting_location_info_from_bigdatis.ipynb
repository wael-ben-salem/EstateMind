````python
data = pd.read_json('datasets/Bigdatis/finale_bigdatis_details.json')
df_bigdatis = json_normalize(data.to_dict('records'))
df_bigdatis.info()
df_bigdatis
````

````python
# 1 — keep one sample per locationId
df_locations = df_bigdatis.drop_duplicates(subset=["locationId"], keep="first")[["locationId", "description"]].reset_index(drop=True)

print(f"Unique locationIds: {len(df_locations)}")
df_locations
````

````python
def extract_location_from_description(description):
    prompt = f"""Extract location information from this real estate description.
Return ONLY a valid JSON object, nothing else.
You can infer fields from context, for example "jardin de carthage" → ville: "Carthage", gouvernerat: "Tunis".

Fields:
- gouvernerat: (string) Tunisian governorate
- ville: (string) city or area
- adresse: (string street or neighborhood
- code_postal: (integer) Tunisian postal code

Description:
{description}

JSON:"""
    try:
        response = requests.post("http://localhost:1234/v1/chat/completions", json={
            "model": "meta-llama-3.1-8b-instruct",
            "messages": [{"role": "user", "content": prompt}],
            "temperature": 0
        })
        raw = response.json()["choices"][0]["message"]["content"].strip()
        raw = raw.replace("```json", "").replace("```", "").strip()
        print(raw)
        return json.loads(raw)
    except Exception as e:
        print(f"Error: {e}")
        return {"gouvernerat": None, "ville": None, "adresse": None, "code_postal": None}

results = []
for idx, row in tqdm(df_locations.iterrows(), total=len(df_locations)):
    print ( row ["description"])
    result = extract_location_from_description(row["description"])
    result["locationId"] = row["locationId"]  # ← dict not list
    results.append(result)

location_map = pd.DataFrame(results)[["locationId", "gouvernerat", "ville", "adresse", "code_postal"]]
location_map.to_csv("location_map.csv", index=False)
location_map.to_json("location_map.json", orient="records", indent=2)

print(f"Saved {len(location_map)} locations")
````
